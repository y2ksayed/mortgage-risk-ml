# Mortgage Risk Analysis

Exploratory analysis of the synthetic mortgage book: distribution of credit scores,
loan-to-value ratios, default rates by state, and the PD model's score calibration.
All data here is simulated - the goal is to validate the pipeline shapes, not to
draw conclusions about real borrowers.

```python
import pandas as pd
import matplotlib.pyplot as plt

apps = pd.read_parquet("../data/sample_applications.parquet")
print(apps.shape)
apps.head()
```

```python
# Default rate by credit tier - expect subprime >> prime
by_tier = apps.groupby("credit_tier")["defaulted"].agg(["mean", "count"])
print(by_tier)
by_tier["mean"].plot(kind="bar", title="Default rate by credit tier")
plt.ylabel("default rate")
plt.tight_layout()
plt.show()
```

```python
# PD score calibration: predicted probability vs realized default rate
scored = pd.read_parquet("../data/sample_scored.parquet")
scored["pd_bucket"] = pd.cut(scored["pd_score"], bins=10)
calib = scored.groupby("pd_bucket", observed=True)["defaulted"].mean()
print(calib)
calib.plot(title="PD calibration: realized default rate per score bucket")
plt.tight_layout()
plt.show()
```
